## Detailed selection

```python
df_hyades_candidates = pd.read_csv('vizier_J_A+A_623_A35/Hyades_rough_selection_with_query_data_gtomo_banyan.csv')
df_praesepe_candidates = pd.read_csv('vizier_J_A+A_628_A66/Praesepe_rough_selection_with_query_data_gtomo_banyan.csv')
df_pleiades_candidates = pd.read_csv('vizier_J_A+A_628_A66/Pleiades_rough_selection_with_query_data_gtomo_banyan.csv')
```

```python
df_hyades_selected = df_hyades_candidates[(df_hyades_candidates['MBP_MRP']<0.5)
                                       & (df_hyades_candidates['Object_Type']!='SB*')
                                       & (df_hyades_candidates['Object_Type']!='BS*')
                                       & (df_hyades_candidates['Object_Type']!='Be*')
                                       & (df_hyades_candidates['RUWE_DR3']<1.4)
                                       & (df_hyades_candidates['P_binary']<0.01)
                                       & (df_hyades_candidates['BANYAN_prob_HYA']>0.5)
]

hyades_exclude_gaia = [
    'Gaia DR2 3312536927686011520' # 80 Tau: close-in binary < 1.7''
]

hyades_exclude_HIP = [
    20156, # high HGCA acceleration
    20648, # high HGCA acceleration
    19261, # high HGCA acceleration
]

hip_id_int_hyades = pd.to_numeric(df_hyades_selected["HIP_ID"], errors="coerce").astype("Int64")

mask_exclude_hyades = (
    df_hyades_selected["target_name"].isin(hyades_exclude_gaia)
    | hip_id_int_hyades.isin(hyades_exclude_HIP)
)

df_hyades_finalized = df_hyades_selected.loc[~mask_exclude_hyades].copy()

df_hyades_finalized.loc[df_hyades_finalized["HD_ID"] == 29499, ["vbroad", "vbroad_err"]] = [86, 0.15 * 86] # Royer+02
df_hyades_finalized.loc[df_hyades_finalized["HD_ID"] == 27934, ["vbroad", "vbroad_err"]] = [94, 0.15 * 95] # Zorec&Royer+12
df_hyades_finalized.loc[df_hyades_finalized["HD_ID"] == 30034, ["vbroad", "vbroad_err"]] = [110, 0.15 * 110] # Zorec&Royer+12
df_hyades_finalized.loc[df_hyades_finalized["HD_ID"] == 27946, ["vbroad", "vbroad_err"]] = [191, 0.15 * 191] # Zorec&Royer+12

df_hyades_finalized.to_csv(DATA_DIR / "clusters" / "Hyades_finalized_targets.csv", index=False)

df_hyades_finalized
```

```python
df_praesepe_selected = df_praesepe_candidates[(df_praesepe_candidates['MBP_MRP']<0.48)
                                            & (df_praesepe_candidates['Object_Type']!='SB*')
                                            & (df_praesepe_candidates['Object_Type']!='BS*')
                                            & (df_praesepe_candidates['Object_Type']!='Be*')
                                            & (df_praesepe_candidates['RUWE_DR3']<1.4)
                                            & (df_praesepe_candidates['P_binary']<0.01)
                                            # & (df_hyades_candidates['BANYAN_prob_PRA']>0.5)
]

praesepe_exclude_gaia = [
    'Gaia DR2 661224581888230400' # V* HI Cnc: close-in binary < 2.2''
]

praesepe_exclude_HD = [
    67959, # high HGCA acceleration
    69477, # high HGCA acceleration
    73175, # high HGCA acceleration
    73819, # high HGCA acceleration
]

hd_id_int_praesepe = pd.to_numeric(df_praesepe_selected["HD_ID"], errors="coerce").astype("Int64")

mask_exclude_praesepe = (
    df_praesepe_selected["target_name"].isin(praesepe_exclude_gaia)
    | hd_id_int_praesepe.isin(praesepe_exclude_HD)
)

df_praesepe_finalized = df_praesepe_selected.loc[~mask_exclude_praesepe].copy()

df_praesepe_finalized.loc[df_praesepe_finalized["HD_ID"] == 73818, ["vbroad", "vbroad_err"]] = [66, 3.3] # Fossati+07
df_praesepe_finalized.loc[df_praesepe_finalized["HD_ID"] == 73854, ["vbroad", "vbroad_err"]] = [68, 3.4] # RACHFORD+1998

updates = {
    # 73499: (), # no available vsini data
    73818: (66, 3.3), # Fossati+07
    73854: (68, 3.4), # RACHFORD+1998
}

df_praesepe_finalized.to_csv(DATA_DIR / "clusters" / "Praesepe_finalized_targets.csv", index=False)

df_praesepe_finalized
```

```python
df_pleiades_selected = df_pleiades_candidates[(df_pleiades_candidates['MBP_MRP'] < 0.18)
                                              & (df_pleiades_candidates['Object_Type'] != 'SB*')
                                              & (df_pleiades_candidates['Object_Type'] != 'BS*')
                                              & (df_pleiades_candidates['Object_Type'] != 'Be*')
                                              & (df_pleiades_candidates['RUWE_DR3'] < 1.4)
                                              & (df_pleiades_candidates['P_binary'] < 0.01)
                                              & (df_pleiades_candidates['BANYAN_prob_PLE']>0.5)
]

pleiades_exclude_gaia = []

pleiades_exclude_HD = [
    23336, # high HGCA acceleration
    23410, # 3.4" binary (VAST; De Rosa et al. 2011)
    23489, # 0.33 arcsec binary https://ui.adsabs.harvard.edu/abs/2020MNRAS.495..806G/abstract
    23568, # high HGCA acceleration
    23155, # high HGCA acceleration
]

hd_id_int_pleiades = pd.to_numeric(df_pleiades_selected["HD_ID"], errors="coerce").astype("Int64")

mask_exclude_pleiades = (
    df_pleiades_selected["target_name"].isin(pleiades_exclude_gaia)
    | hd_id_int_pleiades.isin(pleiades_exclude_HD)
)

df_pleiades_finalized = df_pleiades_selected.loc[~mask_exclude_pleiades].copy()

# df_pleiades_finalized.loc[df_pleiades_finalized["HD_ID"] == 23862, ["vbroad", "vbroad_err"]] = [290, 21] # Zorec+16 # Be star (already removed)
df_pleiades_finalized.loc[df_pleiades_finalized["HD_ID"] == 23753, ["vbroad", "vbroad_err"]] = [317, 25] # Torres+20
df_pleiades_finalized.loc[df_pleiades_finalized["HD_ID"] == 23923, ["vbroad", "vbroad_err"]] = [335, 0.15 * 335] # Zorec&Royer12
df_pleiades_finalized.loc[df_pleiades_finalized["HD_ID"] == 23913, ["vbroad", "vbroad_err"]] = [188, 11] # Torres+20
df_pleiades_finalized.loc[df_pleiades_finalized["HD_ID"] == 23629, ["vbroad", "vbroad_err"]] = [171, 15] # Torres+20
df_pleiades_finalized.loc[df_pleiades_finalized["HD_ID"] == 23402, ["vbroad", "vbroad_err"]] = [285, 11] # Torres+20

df_pleiades_finalized.to_csv(DATA_DIR / "clusters" / "Pleiades_finalized_targets.csv", index=False)

df_pleiades_finalized
```

```python
df_hyades_tmp = df_hyades_finalized.copy()
df_praesepe_tmp = df_praesepe_finalized.copy()
df_pleiades_tmp = df_pleiades_finalized.copy()

df_hyades_tmp.insert(0, "Cluster", "Hyades")
df_praesepe_tmp.insert(0, "Cluster", "Praesepe")
df_pleiades_tmp.insert(0, "Cluster", "Pleiades")

df_all_clusters_saved = pd.concat(
    [
        df_hyades_tmp,
        df_praesepe_tmp,
        df_pleiades_tmp,
    ],
    axis=0,
    ignore_index=True,
)

df_all_clusters_saved.to_csv(DATA_DIR / "clusters" / "targets_for_sampling.csv", index=False)

df_all_clusters_saved
```
